```python
user_steps = activity.groupby("Id")["TotalSteps"].mean()
user_steps
```

```python
def activity_level(steps):
    if steps < 5000:
        return "Low"
    elif steps < 10000:
        return "Moderate"
    else:
        return "High"

user_activity = user_steps.apply(activity_level)

user_activity.value_counts()
```

### Data Analysis Key Findings

- The average number of steps was 6,547 per day.
- Users recorded 10,000 or more steps on 27.8% of days.
- 14 users had low average activity, 15 had moderate activity, and 6 had high activity.
- The average recorded sedentary time was about 995 minutes per day.
- The average sleep duration was about 7 hours per day.
- 44.1% of recorded sleep days had less than 7 hours of sleep.

## 6. Act - Recommendations


### Final Conclusion

The analysis showed that most users had low or moderate activity levels. Only 6 out of 35 users had a high average activity level. The activity data also showed a large amount of sedentary time compared with active minutes.

Sleep was another important finding. Users slept for about 7 hours on average and spent about 7 hours and 38 minutes in bed. Also, 44.1% of recorded sleep days had less than 7 hours of sleep.

Overall, the data show that users could benefit from more daily movement and better sleep habits. Bellabeat could use these insights to encourage users to be more active and support healthier sleep routines through its app.

### Top 3 Recommendations

1. **Encourage daily movement**  
   Bellabeat could use the app to set simple daily step goals and send reminders to users who are less active.

2. **Promote better sleep habits**  
   Bellabeat could provide simple sleep tips and reminders to help users maintain a regular sleep schedule.

3. **Create personalized daily goals**  
   Bellabeat could use activity and sleep data to give users simple daily goals based on their habits.
